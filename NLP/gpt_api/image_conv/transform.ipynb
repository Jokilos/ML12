```python
def delete(text, chars, replace):
    for c, r in zip(chars, replace):
        text = text.replace(c, r)

    return text

def tick_find(line):
    return line.find(tick_marks[0]) != -1 or line.find(tick_marks[1]) != -1

def print_one(text):
    s = text
    answer = None
    for line in s.splitlines():
        if line.find(tick_marks[0]) != -1 or line.find(tick_marks[1]) != -1:
            answer = line

    s = delete(s, ['\n'] + tick_marks, [' ', '', '']) 
    print(s, '---', answer)

for split in text_split:
    count = 0
    for char in split:
        if char in tick_marks:
            count += 1
    marked[split] = count

to_del = []
for s, c in marked.items():
    if c == 0:
        to_del += [s]

for d in to_del:
    del marked[d]

for s, c in marked.items():
    if c == 1:
        print_one(s)
    else:
        start = 0
        for q in range(c):
            lines = s.splitlines()[start:]
            q_text = ''
            tick_found = False

            for line in lines:
                start += 1
                q_text += line + '\n'

                if tick_found and line == '':
                    break
                if tick_found and tick_find(line):
                    assert False, s
                if tick_find(line):
                    tick_found = True
            
            print_one(q_text)
```
