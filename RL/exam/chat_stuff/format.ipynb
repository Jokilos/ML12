````python
import re

with open('chat_log.md', 'r', encoding='utf-8') as f:
    text = f.read()

lines = []
delete = ['Certainly', 'concise', 'let me know']

look_for_term = False
for line in text.splitlines():
    to_del = False
    for d in delete:
        if re.search(d, line, re.IGNORECASE):
            to_del = True

    if look_for_term and line == '```':
        look_for_term = False
        to_del = True

    if re.search('```markdown', line, re.IGNORECASE):
        look_for_term = to_del = True

    if not to_del:
        lines += [line]

topics = [
'# Reinforcement-learning basics',
'# Value-based methods',
'# Function approximation',
'# Policy gradient-methods',
'# Exploration-exploitation',
'# Model-based methods',
'# Advanced topics in RL',
]

line_nums = [] 

for i, line in enumerate(lines):
    for topic in topics:
        if line.find(topic) != -1:
            print(i, line)
            line_nums += [i]

line_nums += [len(lines)]

folder = '../done'
i = 0
for (begin, end) in zip(line_nums[:-1], line_nums[1:]):
    print(begin, end)
    content = '\n'.join(lines[begin:end])

    with open(f'{folder}/{topics[i][2:].replace(' ', '-')}.md', 'w', encoding='utf-8') as f:
        f.write(content)
        # pass
    
    i += 1
````
